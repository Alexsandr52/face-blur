```python
import subprocess
import json
```

```python
video_path = '../data/TATARKA-АЛТЫН-SnapYT.mp4'
```

```python
result = subprocess.run(
    [    'ffprobe',
        '-v', 'error',
        '-show_streams',
        '-show_format',
        '-of', 'json',
        video_path
    ],
    capture_output=True,
    text=True,
    check=True
)

data = json.loads(result.stdout)

video = next(
    s for s in data['streams']
    if s['codec_type'] == 'video'
)

has_audio = any(
    s['codec_type'] == 'audio' 
    for s in data['streams']
)

print("Кодек:", video["codec_name"])
print("Разрешение:", video["width"], "x", video["height"])
print("Средний FPS:", video["avg_frame_rate"])
print("Длительность:", data["format"].get("duration"))
print("Размер файла:", data["format"].get("size"), "байт")
print("Аудио:", has_audio)
```

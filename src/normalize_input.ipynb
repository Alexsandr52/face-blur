```python
import subprocess
import json
```

```python
input_video_file = '../data/TATARKA-АЛТЫН-SnapYT.mp4'
output_video_file = '../data/processed_TATARKA-АЛТЫН-SnapYT.mp4'
```

```python
settings = {
    'fps': 30,
    'video_codec': 'libopenh264',
    'audio_codec': 'aac',
    'pixel_format': 'yuv420p',
    'auto_rotate': True,
}
```

```python
command = [
    'ffmpeg',
    '-autorotate',
    '-i',
    input_video_file,
    '-map',
    '0:v:0',
    '-map',
    '0:a:0?',
    '-vf',
    'fps=30,pad=ceil(iw/2)*2:ceil(ih/2)*2',
    '-c:v',
    settings['video_codec'],
    '-pix_fmt',
    settings['pixel_format'],
    '-c:a',
    settings['audio_codec'],
    '-fps_mode',
    'cfr',
    output_video_file,
]
```

```python
subprocess.run(command, capture_output=True, text=True)
```

```python
import cv2
```

```python
cap = cv2.VideoCapture(output_video_file)

if not cap.isOpened():
    raise RantimeError('Видео не открлось')
```

```python
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = float(cap.get(cv2.CAP_PROP_FPS))
```

```python
print("width:", width, type(width))
print("height:", height, type(height))
print("fps:", fps, type(fps))
```

```python
output_path = "../data/numbered_video.mp4"

fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))
```

```python
if not out.isOpened():
    cap.release()
    raise RuntimeError('Не удалось создать выходное видео')
```

```python
frame_number = 0

try:
    while True:
        ret, frame = cap.read()

        if not ret:
            break

        frame_number += 1

        cv2.putText(
            frame,
            f"Frame: {frame_number}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2,
        )

        out.write(frame)

finally:
    cap.release()
    out.release()

print(f"Обработано кадров: {frame_number}")
```

```python
subprocess.run(
    [
        "ffprobe",
        "-v",
        "error",
        "-show_entries",
        "stream=codec_type",
        "-of",
        "csv=p=0",
        output_path,
    ]
)
```
